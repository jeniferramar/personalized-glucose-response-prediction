### Step 3: Check Whether All 45 CGMacros Files Have the Same Columns

```python
# Check the column structure of every CGMacros file

column_check = {}

for file in cg_files:
    df = pd.read_csv(file, nrows=0)
    column_check[file.name] = df.columns.tolist()

# Use the first file's columns as the expected structure
expected_columns = column_check[cg_files[0].name]

different_files = []

for filename, columns in column_check.items():
    if columns != expected_columns:
        different_files.append(filename)

print("Total CGMacros files checked:", len(column_check))
print("Files with different columns:", different_files)
```

### Step 4: Identify Column Differences Before Combining Files

```python
# Compare every file's columns with CGMacros-001.csv

first_file = cg_files[0]
first_columns = pd.read_csv(first_file, nrows=0).columns.tolist()

print("Reference file:", first_file.name)
print("Reference columns:")
print(first_columns)

for file in cg_files[1:]:
    columns = pd.read_csv(file, nrows=0).columns.tolist()

    if columns != first_columns:
        missing = [col for col in first_columns if col not in columns]
        extra = [col for col in columns if col not in first_columns]

        print("\nFile:", file.name)
        print("Missing compared with first file:", missing)
        print("Extra compared with first file:", extra)
```

### Step 5: Prepare Files Before Combining

```python
# Read and standardize every CGMacros file

cleaned_frames = []

for file in cg_files:
    df = pd.read_csv(file)

    # Remove spaces before and after column names
    df.columns = df.columns.str.strip()

    # Remove unnecessary saved-index columns
    df = df.loc[:, ~df.columns.str.startswith("Unnamed")]

    # Extract subject number from the filename
    subject_id = int(file.stem.split("-")[-1])
    df.insert(0, "subject", subject_id)

    # Keep the source filename for validation
    df.insert(1, "source_file", file.name)

    cleaned_frames.append(df)

print("Files prepared:", len(cleaned_frames))
print("First file shape after preparation:", cleaned_frames[0].shape)
print("First file columns:")
print(cleaned_frames[0].columns.tolist())
```

### Step 6: Combine All 45 CGMacros Files

```python
combined_cg = pd.concat(cleaned_frames, ignore_index=True, sort=False)

print("Combined shape:", combined_cg.shape)
print("Number of subjects:", combined_cg["subject"].nunique())
print("Columns:")
print(combined_cg.columns.tolist())

combined_cg.head()
```

#### Step 11.10 Correcting Out-of-Range Libre GL Values Using Linear Interpolation

```python
import pandas as pd

# ---------------------------------------------------
# Step: Correct Subject 30 Libre GL and Recalculate
# Adjacent 1-minute Values Using Linear Interpolation
# ---------------------------------------------------

subject_id = 30

prev_time = pd.Timestamp("2021-01-24 09:57:00")
mid_time  = pd.Timestamp("2021-01-24 10:12:00")
next_time = pd.Timestamp("2021-01-24 10:27:00")

prev_value = 391
corrected_mid_value = 400
next_value = 386

# IMPORTANT:
# Use the EXISTING column name "Libre GL"
# Do NOT create a new column such as "Libre_Gl"

# First interval: 09:57 to 10:12
mask1 = (
    (combined_cg["subject"] == subject_id) &
    (combined_cg["Timestamp"].between(prev_time, mid_time))
)

mins1 = (
    combined_cg.loc[mask1, "Timestamp"] - prev_time
).dt.total_seconds() / 60

total1 = (
    mid_time - prev_time
).total_seconds() / 60

combined_cg.loc[mask1, "Libre GL"] = (
    prev_value
    + (corrected_mid_value - prev_value) * (mins1 / total1)
).round(2)


# Second interval: 10:12 to 10:27
mask2 = (
    (combined_cg["subject"] == subject_id) &
    (combined_cg["Timestamp"].between(mid_time, next_time))
)

mins2 = (
    combined_cg.loc[mask2, "Timestamp"] - mid_time
).dt.total_seconds() / 60

total2 = (
    next_time - mid_time
).total_seconds() / 60

combined_cg.loc[mask2, "Libre GL"] = (
    corrected_mid_value
    + (next_value - corrected_mid_value) * (mins2 / total2)
).round(2)


# Verify corrected values
display(
    combined_cg.loc[
        (combined_cg["subject"] == subject_id) &
        (combined_cg["Timestamp"].between(prev_time, next_time)),
        ["subject", "Timestamp", "Libre GL"]
    ]
)

# Verify no values remain above documented Libre GL maximum
print(
    "Libre GL values above 400:",
    (combined_cg["Libre GL"] > 400).sum()
)
```
